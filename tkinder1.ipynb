```python
from tkinter import *
def click(event):
    text=event.widget.cget("text")
    if text=="=":
        try:
            result=eval(str(entry.get()))
            entry.delete(0,END)
            entry.insert(END,result)
        except:
            entry.delete(0,END)
            entry.insert(END,"Error")
    elif text=="C":
        entry.delete(0,END)
    else:
        entry.insert(END,text)
root=Tk()
root.geometry("250x350")
root.title("Simple Calculator")
entry=Entry(root, font="Arial 20")
entry.pack(pady=20,fill=BOTH,padx=20)
btns=[
    ['7','8','9','/'],
    ['4','5','6','*'],
    ['1','2','3','-'],
    ['C','0','=','+'],
]
for row in btns:
    frame= Frame(root)
    frame.pack()
    for char in row:
        button=Button(frame,text=char,font="Arial 18",width=4,height=2)
        button.pack(side=LEFT,padx=2,pady=2)
        button.bind("<Button-1>",click)
root.mainloop()
```

```python
import tkinter as tk
from tkinter import ttk, messagebox
from datetime import datetime

# -------------------- PRODUCT DATA --------------------
products = {
    "Grocery": {
        "Rice": {"Basmati Rice": 120, "Brown Rice": 150},
        "Flour": {"Wheat Flour": 55, "Maida": 48},
        "Oil": {"Sunflower Oil": 140, "Mustard Oil": 155, "Olive Oil": 175},
    },
    "Beverages": {
        "Soft Drinks": {"Coke": 45, "Pepsi": 45},
        "Juice": {"Mango Juice": 80, "Orange Juice": 75},
    },
    "Personal Care": {
        "Shampoo": {"Dove Shampoo": 180, "Head & Shoulders": 220},
        "Soap": {"Dove Soap": 55, "Lux Soap": 40},
    },
    "Stationery": {
        "Writing": {"Pen": 10, "Pencil": 5},
        "Books": {"Notebook": 60, "Register": 90},
    }
}

cart = []
bill_no = 1001

# -------------------- FUNCTIONS --------------------
def update_subcategories(event=None):
    category = category_var.get()
    values = list(products.get(category, {}).keys())
    subcategory_combo["values"] = values
    subcategory_var.set("")
    product_combo["values"] = []
    product_var.set("")

def update_products(event=None):
    category = category_var.get()
    subcategory = subcategory_var.get()
    values = list(products.get(category, {}).get(subcategory, {}).keys())
    product_combo["values"] = values
    product_var.set("")

def add_to_cart():
    product = product_var.get()

    if not product:
        messagebox.showwarning("Product", "Please select a product.")
        return

    try:
        quantity = int(quantity_var.get())
        if quantity <= 0:
            raise ValueError
    except ValueError:
        messagebox.showwarning("Quantity", "Enter a valid quantity.")
        return

    category = category_var.get()
    subcategory = subcategory_var.get()
    price = products[category][subcategory][product]

    for item in cart:
        if item["product"] == product:
            item["qty"] += quantity
            refresh_bill()
            clear_product_fields()
            return

    cart.append({"product": product, "qty": quantity, "price": price})
    refresh_bill()
    clear_product_fields()

def remove_from_cart():
    selected = bill_tree.selection()
    if not selected:
        messagebox.showwarning("Remove", "Select an item first.")
        return

    index = int(selected[0])
    del cart[index]
    refresh_bill()

def clear_product_fields():
    category_var.set("")
    subcategory_var.set("")
    product_var.set("")
    quantity_var.set("1")
    subcategory_combo["values"] = []
    product_combo["values"] = []

def clear_cart():
    cart.clear()
    refresh_bill()

def calculate_total():
    total = sum(item["qty"] * item["price"] for item in cart)
    total_var.set(f"₹ {total:.2f}")
    return total

def refresh_bill():
    for item in bill_tree.get_children():
        bill_tree.delete(item)

    for i, item in enumerate(cart):
        amount = item["qty"] * item["price"]
        bill_tree.insert(
            "", "end", iid=str(i),
            values=(
                item["product"],
                item["qty"],
                f"₹ {item['price']:.2f}",
                f"₹ {amount:.2f}"
            )
        )
    calculate_total()

def search_bill():
    number = bill_number_var.get().strip()
    if not number:
        messagebox.showinfo("Search", "Enter a bill number.")
    else:
        messagebox.showinfo("Search", f"Bill Number: {number}")

def generate_bill():
    global bill_no

    customer = customer_name_var.get().strip()
    phone = phone_var.get().strip()

    if not customer:
        messagebox.showwarning("Customer", "Enter customer name.")
        return

    if not cart:
        messagebox.showwarning("Bill", "Add at least one product.")
        return

    current_bill = bill_no
    bill_no += 1
    bill_number_var.set(str(current_bill))

    bill_text.delete("1.0", tk.END)

    bill_text.insert(tk.END, "          BIG BAZAAR (FUTURE RETAIL LTD)\n")
    bill_text.insert(tk.END, "            BB-TANE-DHANSAR THAKUR MALL\n")
    bill_text.insert(tk.END, "                    THAN - 40104\n")
    bill_text.insert(tk.END, "             TEL NO.: 022-28284707\n")
    bill_text.insert(tk.END, "               HELPLINE: 1800-200-255\n")
    bill_text.insert(tk.END, "             GSTIN: 35AABCS1429B1ZX\n")
    bill_text.insert(tk.END, "-" * 68 + "\n")
    bill_text.insert(
        tk.END,
        f"Customer Name: {customer:<22} Phone: {phone}\n"
    )
    bill_text.insert(
        tk.END,
        f"Bill Number: {current_bill:<25} Date: {datetime.now():%d-%m-%Y %H:%M}\n"
    )
    bill_text.insert(tk.END, "-" * 68 + "\n")
    bill_text.insert(
        tk.END,
        f"{'Product Name':<28}{'Qty':>8}{'Price':>14}{'Amount':>18}\n"
    )
    bill_text.insert(tk.END, "-" * 68 + "\n")

    for item in cart:
        amount = item["qty"] * item["price"]
        bill_text.insert(
            tk.END,
            f"{item['product']:<28}{item['qty']:>8}"
            f"{item['price']:>14.2f}{amount:>18.2f}\n"
        )

    total = calculate_total()
    bill_text.insert(tk.END, "-" * 68 + "\n")
    bill_text.insert(tk.END, f"{'TOTAL':>50}{total:>18.2f}\n")
    bill_text.insert(tk.END, "\n             Thank you for shopping with us!\n")

def clear_all():
    customer_name_var.set("")
    phone_var.set("")
    bill_number_var.set("")
    clear_cart()
    bill_text.delete("1.0", tk.END)
    show_empty_bill()

def show_empty_bill():
    bill_text.insert(tk.END, "          BIG BAZAAR (FUTURE RETAIL LTD)\n")
    bill_text.insert(tk.END, "            BB-TANE-DHANSAR THAKUR MALL\n")
    bill_text.insert(tk.END, "                    THAN - 40104\n")
    bill_text.insert(tk.END, "             TEL NO.: 022-28284707\n")
    bill_text.insert(tk.END, "               HELPLINE: 1800-200-255\n")
    bill_text.insert(tk.END, "             GSTIN: 35AABCS1429B1ZX\n\n")
    bill_text.insert(tk.END, "Customer Name:                         Phone Number:\n")
    bill_text.insert(tk.END, "Bill Number:                           Date:\n\n")
    bill_text.insert(tk.END, "Product Name                  Quantity(Qty)       Price\n")
    bill_text.insert(tk.END, "-" * 68 + "\n")

def logout():
    if messagebox.askyesno("Logout", "Do you want to logout?"):
        root.destroy()

def exit_app():
    if messagebox.askyesno("Exit", "Do you want to exit?"):
        root.destroy()

# -------------------- MAIN WINDOW --------------------
root = tk.Tk()
root.title("Billing System")
root.geometry("1500x850")
root.minsize(1100, 700)
root.configure(bg="#ff765f")

# Outer white area
outer = tk.Frame(root, bg="#ff765f")
outer.pack(fill="both", expand=True, padx=22, pady=22)

main = tk.Frame(outer, bg="white")
main.pack(fill="both", expand=True)

# -------------------- HEADER --------------------
header = tk.Frame(main, bg="white", height=120)
header.pack(fill="x", padx=25)
header.pack_propagate(False)

tk.Label(
    header, text="●", bg="white", fg="black",
    font=("Arial", 22)
).place(x=5, y=10)

tk.Button(
    header, text="LOGOUT", command=logout,
    bg="#d91d16", fg="white",
    activebackground="#b71913",
    relief="flat", font=("Arial", 11, "bold"),
    padx=18, pady=7
).place(x=0, y=48)

tk.Label(
    header, text="Billing System",
    bg="white", fg="black",
    font=("Arial", 38, "bold")
).place(relx=0.5, y=18, anchor="n")

# -------------------- CUSTOMER DETAILS --------------------
customer_frame = tk.LabelFrame(
    main, text="Customer Details",
    bg="white", font=("Arial", 12),
    bd=2, relief="groove"
)
customer_frame.pack(fill="x", padx=38, pady=(0, 15))

tk.Label(customer_frame, text="Bill Number", bg="white",
         font=("Arial", 11)).grid(row=0, column=0, padx=12, pady=12)

bill_number_var = tk.StringVar()
tk.Entry(
    customer_frame, textvariable=bill_number_var,
    font=("Arial", 11), width=25, relief="flat",
    highlightthickness=1, highlightbackground="#cccccc"
).grid(row=0, column=1, padx=5, pady=12, sticky="ew")

tk.Button(
    customer_frame, text="Search", command=search_bill,
    bg="#d91d16", fg="white", relief="flat",
    font=("Arial", 11, "bold"), padx=22, pady=6
).grid(row=0, column=2, padx=18)

tk.Label(customer_frame, text="Customer Name", bg="white",
         font=("Arial", 11)).grid(row=0, column=3, padx=10)

customer_name_var = tk.StringVar()
tk.Entry(
    customer_frame, textvariable=customer_name_var,
    font=("Arial", 11), relief="flat",
    highlightthickness=1, highlightbackground="#cccccc"
).grid(row=0, column=4, padx=5, sticky="ew")

tk.Label(customer_frame, text="Contact Number", bg="white",
         font=("Arial", 11)).grid(row=0, column=5, padx=18)

phone_var = tk.StringVar()
tk.Entry(
    customer_frame, textvariable=phone_var,
    font=("Arial", 11), relief="flat",
    highlightthickness=1, highlightbackground="#cccccc"
).grid(row=0, column=6, padx=5, sticky="ew")

customer_frame.columnconfigure(1, weight=1)
customer_frame.columnconfigure(4, weight=1)
customer_frame.columnconfigure(6, weight=1)

# -------------------- CONTENT --------------------
content = tk.Frame(main, bg="white")
content.pack(fill="both", expand=True, padx=38)

left = tk.Frame(content, bg="white", width=560)
left.pack(side="left", fill="both", padx=(0, 8))
left.pack_propagate(False)

right = tk.Frame(content, bg="white")
right.pack(side="left", fill="both", expand=True, padx=(8, 0))

# -------------------- PRODUCTS --------------------
products_frame = tk.LabelFrame(
    left, text="Products",
    bg="white", font=("Arial", 12),
    bd=2, relief="groove"
)
products_frame.pack(fill="both", expand=True)

tk.Label(products_frame, text="Select Category", bg="white",
         font=("Arial", 11)).pack(anchor="w", padx=15, pady=(25, 5))

category_var = tk.StringVar()
category_combo = ttk.Combobox(
    products_frame, textvariable=category_var,
    values=list(products.keys()), state="readonly"
)
category_combo.pack(fill="x", padx=15)
category_combo.bind("<<ComboboxSelected>>", update_subcategories)

tk.Label(products_frame, text="Sub Category", bg="white",
         font=("Arial", 11)).pack(anchor="w", padx=15, pady=(16, 5))

subcategory_var = tk.StringVar()
subcategory_combo = ttk.Combobox(
    products_frame, textvariable=subcategory_var,
    state="readonly"
)
subcategory_combo.pack(fill="x", padx=15)
subcategory_combo.bind("<<ComboboxSelected>>", update_products)

tk.Label(products_frame, text="Product", bg="white",
         font=("Arial", 11)).pack(anchor="w", padx=15, pady=(16, 5))

product_var = tk.StringVar()
product_combo = ttk.Combobox(
    products_frame, textvariable=product_var,
    state="readonly"
)
product_combo.pack(fill="x", padx=15)

tk.Label(products_frame, text="Quantity", bg="white",
         font=("Arial", 11)).pack(anchor="w", padx=15, pady=(16, 5))

quantity_var = tk.StringVar(value="1")
ttk.Combobox(
    products_frame, textvariable=quantity_var,
    values=[str(i) for i in range(1, 21)],
    state="readonly"
).pack(fill="x", padx=15)

button_row = tk.Frame(products_frame, bg="white")
button_row.pack(side="bottom", pady=30)

def red_button(parent, text, command):
    return tk.Button(
        parent, text=text, command=command,
        bg="#d91d16", fg="white",
        activebackground="#b71913",
        relief="flat", font=("Arial", 11, "bold"),
        padx=18, pady=7, cursor="hand2"
    )

red_button(button_row, "Add to Cart", add_to_cart).pack(side="left", padx=4)
red_button(button_row, "Remove", remove_from_cart).pack(side="left", padx=4)
red_button(button_row, "Clear", clear_cart).pack(side="left", padx=4)

# -------------------- BILL OPTIONS --------------------
options = tk.LabelFrame(
    left, text="Bill Options",
    bg="white", font=("Arial", 12),
    bd=2, relief="groove"
)
options.pack(fill="x", pady=(10, 0))

red_button(options, "Total", calculate_total).pack(side="left", padx=10, pady=18)
red_button(options, "Generate", generate_bill).pack(side="left", padx=10, pady=18)
red_button(options, "Clear", clear_all).pack(side="left", padx=10, pady=18)
red_button(options, "Exit", exit_app).pack(side="left", padx=10, pady=18)

# -------------------- BILL WINDOW --------------------
bill_frame = tk.LabelFrame(
    right, text="Bill Window",
    bg="white", font=("Arial", 12),
    bd=2, relief="groove"
)
bill_frame.pack(fill="both", expand=True)

bill_text = tk.Text(
    bill_frame, font=("Courier New", 10),
    bg="white", fg="black", relief="flat",
    wrap="none"
)
bill_text.pack(fill="both", expand=True, padx=15, pady=15)

# Item selection table
bill_tree = ttk.Treeview(
    right,
    columns=("product", "qty", "price", "amount"),
    show="headings", height=5
)
bill_tree.heading("product", text="Product")
bill_tree.heading("qty", text="Qty")
bill_tree.heading("price", text="Price")
bill_tree.heading("amount", text="Amount")
bill_tree.column("product", width=230)
bill_tree.column("qty", width=70, anchor="center")
bill_tree.column("price", width=100, anchor="e")
bill_tree.column("amount", width=110, anchor="e")
bill_tree.pack(fill="x", padx=10, pady=(5, 8))

total_bar = tk.Frame(right, bg="white")
total_bar.pack(fill="x", padx=12, pady=(0, 10))

tk.Label(
    total_bar, text="Grand Total:",
    bg="white", font=("Arial", 13, "bold")
).pack(side="right", padx=8)

total_var = tk.StringVar(value="₹ 0.00")
tk.Label(
    total_bar, textvariable=total_var,
    bg="white", fg="#d91d16",
    font=("Arial", 14, "bold")
).pack(side="right")

show_empty_bill()

root.mainloop()
```
